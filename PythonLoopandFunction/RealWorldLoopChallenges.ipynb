```python
transactions = [1200, 450, 800, 1500, 2300, 700, 100]
#calculate total transaction value without using sum().
total_transaction = 0
for transaction in transactions:
    total_transaction += transaction
print("Total transaction value:", total_transaction)    
```

```python
#From the same list, find the highest and lowest transaction without max() and min().
highest_transaction = transactions[0]
lowest_transaction = transactions[0]

for transaction in transactions:
    if transaction > highest_transaction:
        highest_transaction = transaction
    else:
        highest_transaction    
    if transaction < lowest_transaction:
        lowest_transaction = transaction
    else:
        lowest_transaction    

print("Highest transaction:", highest_transaction)
print("Lowest transaction:", lowest_transaction)
```

```python
temperatures = [32, 35, 28, 40, 38, 31, 42]
temp_sum = 0
for temp in temperatures:
    temp_sum += temp
print("Sum of temperatures:", temp_sum)
avg_temperature = temp_sum / len(temperatures)
print("Average temperature:", avg_temperature)
```

```python
marks = [78, 92, 45, 67, 88, 53, 99]
cnt_above_90 = 0
cnt_75_89 = 0
cnt_50_74 = 0
cnt_below_50 = 0
for mark in marks:
    if mark > 90:
        cnt_above_90 += 1
    elif mark >= 75 and mark <= 89:
        cnt_75_89 += 1
    elif mark >= 50 and mark <= 74:
        cnt_50_74 += 1
    else:
        cnt_below_50 += 1

print("Number of students with marks above 90:", cnt_above_90)
print("Number of students with marks between 75 and 89:", cnt_75_89)
print("Number of students with marks between 50 and 74:", cnt_50_74)
print("Number of students with marks below 50:", cnt_below_50)
```

```python
#Create a simple login system with a maximum of 3 password attempts
correct_password = "secret123"
max_attempts = 3
attempts = 0
while attempts < max_attempts:
    entered_password = input("Enter your password: ")
    if entered_password == correct_password:
        print("Login successful!")
        break
    else:
        attempts += 1
        print(f"Incorrect password. You have {max_attempts - attempts} attempts left.")
else:
    print("Too many failed attempts. Access denied.")
```

```python
#Count the frequency of every character in a string without using Counter.
word = input("Enter a string: ")
char_frequency = {}
for char in word:
    if char in char_frequency:
        char_frequency[char] += 1
    else:
        char_frequency[char] = 1
print("Character frequencies:")
for char, freq in char_frequency.items():
    print(f"'{char}': {freq}")
```

```python
#Find the second-largest number in a list without using sort().
lst_of_numbers = [12, 45, 23, 67, 34, 89, 90, 56]
largest = second_largest = float('-inf')

for num in lst_of_numbers:
    if num > largest:
        second_largest = largest
        largest = num
        print(largest, second_largest)
    elif num > second_largest and num != largest:
        second_largest = num

print("The second-largest number is:", second_largest)
```

```python
#Check whether a string is a palindrome using loops.
word = input("Enter a string: ")
cleaned_word = ''.join(c.lower() for c in word if c.isalnum())
is_palindrome = True
for i in range(len(cleaned_word) // 2):
    if cleaned_word[i] != cleaned_word[-(i + 1)]:
        is_palindrome = False
        break
if is_palindrome:
    print("The string is a palindrome.")
else:
    print("The string is not a palindrome.")
```
