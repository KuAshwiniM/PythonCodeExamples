A Python dictionary is a built-in, mutable data structure that stores data collections in key: value pairs. Unlike sequences like lists or tuples that use numeric indexes, dictionaries use unique keys to look up values instantly, acting like a hash map or an associative array.

Key Characteristics
Key-Value Mapping:  Every entry consists of a key and a corresponding value.
Ordered:  Starting in Python 3.7, dictionaries maintain the exact order in which items are inserted.
Unique Keys:  Duplicate keys are not allowed. Assigning a value to an existing key overwrites the old value.
Key Restrictions:  Keys must be of an immutable data type, such as strings, numbers, or tuples.
Flexible Values:  Values can be of any data type, including lists, booleans, or even other dictionaries.

```python
#Dictionary creation
student = {
"name": "Rahul",
"age": 22,
"course": "Python",
"city": "Bangalore",
"marks": 88}
student
```

```python
print(student['name'])
print(student['course'])
```

```python
print(student.keys())
print(student.values())
```

```python
for key, value in student.items():
    print(key +":"+ str(value))
```

```python
student["marks"] = 92
student
```

```python
student["email"]="student@gmail.com"
student["phone"]= 1234567890
student
```

```python
student.pop("city")
student
```

```python
student.get("name")
```
