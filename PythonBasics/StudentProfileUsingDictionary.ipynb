A Python dictionary is a built-in data structure used to store data as key-value pairs, where each unique key maps to a specific value. It acts like a real-world dictionary: you look up a specific "word" (the key) to find its "definition" (the value). Dictionaries are optimized for extremely fast data retrieval because they are implemented as hash tables behind the scenes.

Key CharacteristicsOrdered: 
Dictionaries preserve the exact insertion order of items.
Mutable: You can freely add, modify, or remove items after creation.
Unique Keys: Duplicate keys are not allowed; assigning a value to an existing key overwrites the old value.
Key Requirements: Keys must be of an immutable data type (like strings, integers, or tuples).
Flexible Values: Values can be absolutely anything, including lists, integers, strings, or even other dictionaries.

```python
student = {
"name": "Rahul",
"age": 22,
"course": "Python",
"city": "Bangalore",
"marks": 88
}
student
```

```python
print(student['name'])
print(student['course'])
```

```python
print(student.keys())
print(student.values())
```

```python
for key, value in student.items():
    print(key +":"+ str(value))
```

```python
student["marks"] = 92
student
```

```python
student["email"]="student@gmail.com"
student["phone"]= 1234567890
student
```

```python
student.pop("city")
student
```

```python
student.get("name")
```
