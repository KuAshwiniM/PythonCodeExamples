In Python, a tuple is a built-in data type used to store collections of data. Tuple is immutable (cannot be changed, added to, or modified after creation).Tuples are typically written with round brackets () and elements are separated by commas.

Key Characteristic
Immutability - You cannot change, add, or remove elements once defined. Attempting to do so raises a TypeError.
Heterogeneous - Can store multiple data types at once (e.g., a mix of strings, integers, and lists).
Allows Duplicates - Duplicate values are allowed and retain their individual index locations.

```python
technologies = ("Python", "Java", "Python", "C++", "JavaScript", "Python")
print(technologies)
print(type(technologies))
```

```python
#Print the first item
print(technologies[0])

#Print the last item
print(technologies[-1])

#Slice the tuple
print(technologies[1:4])
```

```python
print(technologies.count("Python"))
```

```python
#Find the index of "C++"
print(technologies.index("C++"))

#Find the tuple length
print(len(technologies))

#Convert the tuple into a list
technologies_list = list(technologies)
print(technologies_list)

#Add "Go" after converting it into a list
technologies_list.append("Go")
print(technologies_list)

#Convert it back into a tuple
technologies = tuple(technologies_list)
print(technologies)
```
